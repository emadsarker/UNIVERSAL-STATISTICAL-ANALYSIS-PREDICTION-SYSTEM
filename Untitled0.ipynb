```python
# the statiscal form of app
import io
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# the third termial of the code
from google.colab import files
from IPython.display import display, Markdown
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    accuracy_score, balanced_accuracy_score, classification_report
)

print("UNIVERSAL STATISTICAL ANALYSIS & PREDICTION SYSTEM")
print("=" * 55)

# 1. Upload CSV or Excel data
uploaded = files.upload()

if not uploaded:
    raise ValueError("Please upload a CSV or Excel file.")

filename = next(iter(uploaded))

if filename.lower().endswith(".csv"):
    try:
        df = pd.read_csv(io.BytesIO(uploaded[filename]))
    except UnicodeDecodeError:
        df = pd.read_csv(
            io.BytesIO(uploaded[filename]), encoding="latin-1"
        )
elif filename.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(io.BytesIO(uploaded[filename]))
else:
    raise ValueError("Supported formats: CSV, XLSX, XLS.")

# Clean completely empty columns and normalize column names
df.columns = [
    str(c).strip() if str(c).strip() else f"Column_{i+1}"
    for i, c in enumerate(df.columns)
]
df = df.dropna(axis=1, how="all")

if df.empty:
    raise ValueError("The uploaded dataset is empty.")

print("\n1. DATASET OVERVIEW")
print("-" * 40)
print("Rows:", df.shape[0])
print("Variables:", df.shape[1])
print("Column names:", list(df.columns))
display(df.head())

# 2. Data types and missing values
print("\n2. DATA QUALITY")
quality = pd.DataFrame({
    "Data type": df.dtypes.astype(str),
    "Missing": df.isna().sum(),
    "Missing (%)": (df.isna().mean() * 100).round(2),
    "Unique values": df.nunique(dropna=True)
})
display(quality)

duplicates = int(df.duplicated().sum())
print("Exact duplicate rows:", duplicates)

# 3. Descriptive statistics
numeric = df.select_dtypes(include=np.number)

print("\n3. DESCRIPTIVE STATISTICS")
if not numeric.empty:
    summary = numeric.describe().T
    summary["median"] = numeric.median()
    summary["skewness"] = numeric.skew()
    summary["variance"] = numeric.var()
    display(summary)

    # Visualize numeric distributions
    for col in numeric.columns[:8]:
        values = numeric[col].dropna()
        if len(values):
            plt.figure(figsize=(6, 3))
            plt.hist(values, bins="auto", edgecolor="black")
            plt.title(f"Distribution: {col}")
            plt.xlabel(col)
            plt.ylabel("Frequency")
            plt.tight_layout()
            plt.show()
else:
    print("No numeric columns found.")

# 4. Correlation analysis
print("\n4. CORRELATION ANALYSIS")
if numeric.shape[1] >= 2:
    corr = numeric.corr()
    display(corr.style.format("{:.3f}"))

    print("\nStrong numeric associations (|r| >= 0.70):")
    found = False
    cols = corr.columns.tolist()
    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):
            r = corr.iloc[i, j]
            if pd.notna(r) and abs(r) >= 0.70:
                print(f"{cols[i]} and {cols[j]}: r = {r:.3f}")
                found = True
    if not found:
        print("No strong pairwise linear correlations found.")
    print("Remember: correlation does not prove causation.")
else:
    print("At least two numeric columns are needed.")

# 5. Automatic preliminary recommendations
print("\n5. STATISTICAL DECISION SUPPORT")
if df.isna().sum().sum():
    print("- Investigate missing values before choosing an imputation method.")
else:
    print("- No missing values detected.")

if duplicates:
    print(f"- Review {duplicates} duplicate rows before removing them.")

for col in numeric.columns:
    s = numeric[col].dropna()
    if len(s) >= 8:
        q1, q3 = s.quantile([0.25, 0.75])
        iqr = q3 - q1
        if iqr > 0:
            count = int(
                ((s < q1 - 1.5 * iqr) |
                 (s > q3 + 1.5 * iqr)).sum()
            )
            if count:
                print(f"- Investigate {count} potential outliers in '{col}'.")
#the first part
# 6. Predict an outcome selected by the user
print("\n6. PREDICTION")
candidates = [
    c for c in df.columns
    if df[c].notna().sum() >= 12
    and df[c].nunique(dropna=True) > 1
]

if not candidates:
    print("Not enough usable observations or variation for prediction.")
else:
    print("Available target variables:")
    for i, col in enumerate(candidates, 1):
        print(f"{i}. {col}")

    choice = input(
        "\nEnter the number of the variable you want to predict "
        "(or press Enter to skip): "
    ).strip()

    if choice:
        try:
            target = candidates[int(choice) - 1]
        except (ValueError, IndexError):
            raise ValueError("Please enter a valid target number.")

        work = df.dropna(subset=[target]).copy()
        X = work.drop(columns=[target])
        y = work[target]

        # Remove identifications of the major part..
        X = X.loc[:, [
            c for c in X.columns
            if X[c].notna().any() and X[c].nunique(dropna=True) > 1
        ]]

        # Infer command of the resources
        if pd.api.types.is_numeric_dtype(y):
            if (
                y.nunique() <= 10
                and pd.api.types.is_integer_dtype(y)
            ):
                task = input(
                    "Target is integer-valued. Type C for classification "
                    "or R for regression: "
                ).strip().upper()
                task = "classification" if task == "C" else "regression"
            else:
                task = "regression"
        else:
            task = "classification"

        if X.shape[1] == 0:
            print("No useful predictor columns remain.")
        elif len(work) < 12:
            print("Use more observations for a basic train/test evaluation.")
        elif task == "classification" and y.nunique() < 2:
            print("Classification requires at least two classes.")
        elif task == "classification" and y.value_counts().min() < 2:
            print("Some classes have fewer than two observations. "
                  "More data is needed for stratified evaluation.")
        else:
            numeric_cols = X.select_dtypes(include=np.number).columns.tolist()
            categorical_cols = [
                c for c in X.columns if c not in numeric_cols
            ]

            preprocessing = ColumnTransformer([
                ("num", Pipeline([
                    ("imputer", SimpleImputer(strategy="median")),
                    ("scale", StandardScaler())
                ]), numeric_cols),
                ("cat", Pipeline([
                    ("imputer", SimpleImputer(strategy="most_frequent")),
                    ("encode", OneHotEncoder(handle_unknown="ignore"))
                ]), categorical_cols)
            ])

            stratify = y if task == "classification" else None

            X_train, X_test, y_train, y_test = train_test_split(
                X, y, test_size=0.20, random_state=42,
                stratify=stratify
            )

            if task == "regression":
                model = RandomForestRegressor(
                    n_estimators=200, random_state=42
                )
            else:
                model = RandomForestClassifier(
                    n_estimators=200, random_state=42,
                    class_weight="balanced"
                )

            pipeline = Pipeline([
                ("preprocessing", preprocessing),
                ("model", model)
            ])

            pipeline.fit(X_train, y_train)
            predictions = pipeline.predict(X_test)

            print(f"\nTarget: {target}")
            print(f"Prediction type: {task}")
            print(f"Training rows: {len(X_train)}")
            print(f"Test rows: {len(X_test)}")

            if task == "regression":
                mae = mean_absolute_error(y_test, predictions)
                rmse = np.sqrt(mean_squared_error(y_test, predictions))
                r2 = r2_score(y_test, predictions)

                print(f"Mean Absolute Error: {mae:.4f}")
                print(f"Root Mean Squared Error: {rmse:.4f}")
                print(f"R-squared: {r2:.4f}")

                if r2 < 0:
                    print("DECISION: Performance is worse than the "
                          "usual test-mean baseline. Do not rely on it.")
                elif r2 < 0.30:
                    print("DECISION: Predictive performance may be weak. "
                          "Review data and predictors.")
                else:
                    print("DECISION: Review the prediction errors and "
                          "real-world costs before using the model.")

            else:
                acc = accuracy_score(y_test, predictions)
                bal = balanced_accuracy_score(y_test, predictions)

                print(f"Accuracy: {acc:.4f}")
                print(f"Balanced accuracy: {bal:.4f}")
                print(classification_report(
                    y_test, predictions, zero_division=0
                ))

                print("DECISION: Check class-specific errors and the "
                      "cost of false positives and false negatives.")

            results = pd.DataFrame({
                "Actual": np.asarray(y_test),
                "Predicted": predictions
            })
            print("\nActual versus predicted values:")
            display(results.head(50))

            print(
                "\nCAUTION: Results depend on sample size, data quality, "
                "target selection, and test-set representativeness. "
                "Predictive association does not prove causation."
            )
    else:
        print("Prediction skipped. Statistical analysis is complete.")

print("\nAnalysis finished.")
```
